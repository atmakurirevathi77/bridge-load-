```python
"""
PROJECT OBJECTIVE
-----------------
Three supports carry a bridge.

Loads act at three points on the bridge.
We represent the simplified structural model as:

                    A X = B

where:
    A = 3x3 bridge/support coefficient matrix
    X = unknown reaction-force vector
    B = load vector

The system is solved using:
    1. Matrix inverse
    2. Direct linear solver

Then one load is changed to simulate a truck moving,
and the new support reactions are calculated.

IMPORTANT:
This is an educational mathematical model, NOT a replacement
for professional structural-engineering software.
"""
```

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

# Make NumPy output easier to read
np.set_printoptions(precision=3, suppress=True)

print("Libraries imported successfully.")
```

```python
A = np.array([
    [4.0, 1.0, 0.5],
    [1.0, 5.0, 1.0],
    [0.5, 1.0, 4.0]
], dtype=float)


# ------------------------------------------------------------
# Original loads at three points
# ------------------------------------------------------------

loads_before = np.array([
    100.0,   # Load at Point 1 (kN)
    150.0,   # Load at Point 2 (kN)
    80.0     # Load at Point 3 (kN)
], dtype=float)


# ------------------------------------------------------------
# Truck movement:
# Change the load at Point 2
# ------------------------------------------------------------

loads_after = np.array([
    100.0,   # Point 1 remains unchanged
    220.0,   # Truck moves here / load increases
    80.0     # Point 3 remains unchanged
], dtype=float)


print("Bridge model configured.")
print("\nMatrix A:")
print(A)

print("\nLoads before truck movement:")
print(loads_before)

print("\nLoads after truck movement:")
print(loads_after)
```

```python
def validate_inputs(A, loads):
    """Check whether the mathematical inputs are valid."""

    A = np.asarray(A, dtype=float)
    loads = np.asarray(loads, dtype=float)

    # Check matrix dimensions
    if A.shape != (3, 3):
        raise ValueError("Matrix A must be 3 x 3.")

    # Check load vector
    if loads.shape != (3,):
        raise ValueError("Load vector must contain exactly 3 values.")

    # Check numerical values
    if not np.all(np.isfinite(A)):
        raise ValueError("Matrix A contains invalid values.")

    if not np.all(np.isfinite(loads)):
        raise ValueError("Loads contain invalid values.")

    # Check determinant
    determinant = np.linalg.det(A)

    if abs(determinant) < 1e-10:
        raise ValueError(
            "Matrix A is singular or nearly singular. "
            "A unique solution cannot be guaranteed."
        )

    return True


validate_inputs(A, loads_before)
validate_inputs(A, loads_after)

print("✓ Inputs are valid.")
print(f"✓ det(A) = {np.linalg.det(A):.4f}")
print("✓ A unique solution exists.")
```

```python
print("=" * 60)
print("MATHEMATICAL MODEL")
print("=" * 60)

print("\nWe use the linear system:")
print("        A X = B")

print("\nA =")
print(A)

print("\nX =")
print("[R1]")
print("[R2]")
print("[R3]")

print("\nB =")
print(loads_before)

print("\nTherefore:")
print("A × [R1, R2, R3]^T = B")

print("\nWhere:")
print("R1 = reaction at Support 1")
print("R2 = reaction at Support 2")
print("R3 = reaction at Support 3")
```

```python
A_sym = sp.Matrix(A)

det_A = A_sym.det()
A_inverse = A_sym.inv()

print("Determinant of A:")
print(det_A)

print("\nMatrix inverse A⁻¹:")
display(A_inverse)
```
